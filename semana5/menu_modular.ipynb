Parte A — Opción 1: Tuplas


```python
def sumar_tupla(tupla):
    total = 0
    for n in tupla:
        total = total + n
    return total

def seccion_tuplas():
    print("\n--- OPCIÓN 1: TUPLAS ---")
    
    numeros = (4, 8, 2, 10, 5)
    print("Tupla original:", numeros)
   
    print("Tercer elemento:", numeros[2])
    

    num1 = int(input("Ingresa el primer número: "))
    num2 = int(input("Ingresa el segundo número: "))

    nueva_tupla = numeros + (num1, num2)
    print("Nueva tupla:", nueva_tupla)

    lista_ordenada = list(nueva_tupla)
    lista_ordenada.sort()
    print("Lista ordenada:", lista_ordenada)
    
  
    resultado_suma = sumar_tupla(nueva_tupla)
    print("La suma total de la nueva tupla es:", resultado_suma)
seccion_tuplas()
```

Parte B — Opción 2: Diccionarios

```python
def buscar_telefono(agenda, nombre_buscar):
    if nombre_buscar in agenda:
        return agenda[nombre_buscar]
    else:
        return f"{nombre_buscar} no está en tus contactos"

def seccion_diccionarios():
    print("\n--- OPCIÓN 2: DICCIONARIOS ---")
    contactos = {
        "Majo": "442-111-2233",
        "Diego": "442-444-5566",
        "Gohan": "442-777-8899"
    }
    
    nuevo_nombre = input("ingrese el nombre del nuevo contacto: ").strip().title()
    nuevo_tel = input("Ingresa el teléfono del nuevo contacto: ").strip()
    contactos[nuevo_nombre] = nuevo_tel
    
    print("Los nombres registrados son:")
    for nombre in contactos.keys():
        print("-", nombre)
        
    buscar = input("\nEscribe el nombre a buscar: ").strip().title()
    resultado = buscar_telefono(contactos, buscar)
    print("Resultado:", resultado)

seccion_diccionarios()
```

Parte C — Opción 3: Excepciones

```python
def seccion_excepciones():
    print("\n--- OPCIÓN 3: EXCEPCIONES ---")
    try:
        a = int(input("Ingresa el primer número: "))
        b = int(input("Ingresa el segundo número: "))
        
        suma = a + b
        division = a / b
        print("Suma:", suma)
        print("División:", division)
        
    except ValueError:
        print("Eerror, escribe solo numeros enteros, tampoco puedes poner solo un espacio en blanclop")
    except ZeroDivisionError:
        print("No se puede dividir entre cero, intenta otra ves")

seccion_excepciones()
```

Parte D — Opción 4: Strings

```python
def contar_palabras(texto):
    palabras = texto.split()
    return len(palabras)

def seccion_strings():
    print("\n--- OPCIÓN 4: STRINGS ---")

    mensaje = "ya me cansé de hacer esta tarea"
    print("El mensaje original es:", mensaje)
    
    print("La longitud del mensaje es:", len(mensaje))
    
    print("El mensaje en mayúsculas es:", mensaje.upper())
    
    mensaje_nuevo = mensaje.replace("python", "casa")
    print("El mensaje modificado es:", mensaje_nuevo)
    
    total_palabras = contar_palabras(mensaje)
    print("La cantidad de palabras es:", total_palabras)

seccion_strings()
```

Parte E — El menú

```python
while True:
    print("\n=========================================")
    print("1) Tuplas  2) Diccionarios  3) Excepciones  4) Strings  5) Salir")
    opcion = input("Elige una de las opciones del menú: ").strip()
    
    if opcion == "1":
        seccion_tuplas()
    elif opcion == "2":
        seccion_diccionarios()
    elif opcion == "3":
        seccion_excepciones()
    elif opcion == "4":
        seccion_strings()
    elif opcion == "5":
        print("Se cierra el programa! Adiós")
        break
    else:
        print("error, elige una opcion que si esté en el menú")
```
