Ejercicio 1 — Bitácora de la semana

```python
dias = ("Lunes", "Martes", "Miércoles", "Jueves", "Viernes", "Sábado", "Domingo")
km = (120, 85, 0, 210, 95, 0, 150)

def resumen(km):
    total = sum(km)
    dias_rodados = 0
    suma_rodados = 0
    max_km = -1
    idx_max = 0
    for i in range(len(km)):
        if km[i] > 0:
            dias_rodados += 1
            suma_rodados += km[i]
        if km[i] > max_km:
            max_km = km[i]
            idx_max = i
    promedio = round(suma_rodados / dias_rodados, 1) if dias_rodados > 0 else 0.0
    return total, dias_rodados, promedio, idx_max

total, dias_rodados, promedio, idx_max = resumen(km)
print(f"Total: {total} km")
print(f"Días con rodada: {dias_rodados}")
print(f"Promedio por día rodado: {promedio} km")
print(f"Día más largo: {dias[idx_max]} ({km[idx_max]} km)")
```

Ejercicio 3 — Tramos de la ruta a la Sierra

```python
paradas = (("Querétaro", 0), ("San Juan del Río", 52), ("Tequisquiapan", 74), ("Bernal", 125), ("Jalpan", 260))

max_dist = -1
tramo_largo = ""
for i in range(len(paradas) - 1):
    origen, k1 = paradas[i]
    destino, k2 = paradas[i+1]
    dist = k2 - k1
    print(f"{origen} → {destino}: {dist} km")
    if dist > max_dist:
        max_dist = dist
        tramo_largo = f"{origen} → {destino}"

print(f"Tramo más largo: {tramo_largo} ({max_dist} km)")
```

Ejercicio 4 — ¿Con chamarra o sin chamarra?

```python
temps = (8, 14, 22, 27, 31, 19, 11)

def clasificar(t):
    if t < 12:
        return "frío"
    elif t <= 24:
        return "templado"
    else:
        return "calor"

etiquetas = []
contador = {"frío": 0, "templado": 0, "calor": 0}

for t in temps:
    etiqueta = clasificar(t)
    etiquetas.append(etiqueta)
    contador[etiqueta] += 1

# Encontrar cuál predomina
predomina = ""
max_v = -1
for k, v in contador.items():
    if v > max_v:
        max_v = v
        predomina = k

print(etiquetas)
print(contador)
print("Predomina:", predomina)
```

Ejercicio 5 — ¿Cuánto dura el show?

```python
setlist = (("Intro", 95), ("Riff asesino", 245), ("Balada", 302), ("Solo", 128), ("Encore", 409))

def formato(seg):
    m = seg // 60
    s = seg % 60
    if s < 10:
        return f"{m}:0{s}"
    else:
        return f"{m}:{s}"

total_seg = 0
for cancion, seg in setlist:
    total_seg += seg
    print(f"{cancion}: {formato(seg)}")

print(f"Duración total: {formato(total_seg)}")
```

Ejercicio 7 — Boleta del grupo

```python
notas = {"Ana": [90, 85, 78], "Luis": [60, 72, 65], "Mía": [100, 95, 98], "Beto": [70, 68, 74]}

def promedio(lista):
    suma = 0
    for n in lista:
        suma += n
    return round(suma / len(lista), 1)

promedios = {}
mejor_promedio = -1
mejor_alumno = ""

for alumno, lista_cal in notas.items():
    p = promedio(lista_cal)
    promedios[alumno] = p
    
    if p >= 70:
        estado = "Aprobado"
    else:
        estado = "Reprobado"
        
    print(f"{alumno}: {p} - {estado}")
    
    if p > mejor_promedio:
        mejor_promedio = p
        mejor_alumno = alumno

print(f"Mejor promedio: {mejor_alumno}")
```

Ejercicio 10 — Traductor de mensajes de WhatsApp

```python
abrev = {"q": "que", "xq": "porque", "tmb": "también", "pq": "porque", "bn": "bien", "tqm": "te quiero mucho"}

def expandir(mensaje):
    palabras = mensaje.lower().split()
    traducidas = []
    for p in palabras:
        traducidas.append(abrev.get(p, p))
    return " ".join(traducidas)

print(expandir("Q onda tmb voy al concierto xq tocan bn"))
print(expandir("tqm Mamá"))
```

Ejercicio 13 — Capturas sucias

```python
capturas = ["10", "7", "x", "", "9", "8.5", " 6 "]

validas = []
invalidas = 0

for item in capturas:
    try:
        validas.append(int(item))
    except ValueError:
        invalidas += 1

prom = round(sum(validas) / len(validas), 2) if validas else 0.0
print(f"Válidas: {validas}")
print(f"Inválidas: {invalidas}")
print(f"Promedio: {prom}")
```

Ejercicio 16 — ¿Tu contraseña aguanta?

```python
def evaluar(pw):
    fallas = []
    if len(pw) < 8:
        fallas.append("mínimo 8 caracteres")
    
    num = False
    mayus = False
    espacio = False
    for c in pw:
        if c.isdigit():
            num = True
        if c.isupper():
            mayus = True
        if c == " ":
            espacio = True
            
    if not num:
        fallas.append("falta un número")
    if not mayus:
        fallas.append("falta una mayúscula")
    if espacio:
        fallas.append("sin espacios")
        
    return fallas if fallas else "segura"

for pw in ["metal", "Harley2008", "road glide 08", "KLR650moto"]:
    print(f"{pw}: {evaluar(pw)}")
```

Ejercicio 18 — Contador de vocales

```python
frase = "el rock nunca va a morir en esta ciudad"

def vocales(palabra):
    cont = 0
    for c in palabra:
        if c in "aeiouáéíóú":
            cont += 1
    return cont

resultado = {}
max_v = -1
palabra_max = ""

for p in frase.split():
    v = vocales(p)
    resultado[p] = v
    if v > max_v:
        max_v = v
        palabra_max = p

print(resultado)
print(f"Más vocales: {palabra_max} ({max_v})")
```

Ejercicio 20 — El censor del salón

```python
prohibidas = ["tonto", "menso", "baboso"]

def censurar(frase):
    palabras = frase.split()
    censuradas = 0
    nuevas = []
    for p in palabras:
        if p.lower() in prohibidas:
            nuevas.append("*" * len(p))
            censuradas += 1
        else:
            nuevas.append(p)
    return " ".join(nuevas), censuradas

for f in ["no seas menso y pásame el código", "Tonto el que no haga commit", "todo bien profe"]:
    texto, cant = censurar(f)
    print(f"{texto}  ({cant} censuradas)")
```

Ejercicio 21 — Estacionamiento del foro

```python
autos = [("ABC123", 9, 11), ("XYZ789", 8, 15), ("MOT650", 10, 11), ("QRO001", 13, 13)]

def cobro(entrada, salida):
    horas = salida - entrada
    if horas <= 0:
        pago = 20
    else:
        pago = 20 + (horas - 1) * 15
    return horas, pago

dic_cobros = {}
total_dia = 0

for placa, ent, sal in autos:
    h, p = cobro(ent, sal)
    dic_cobros[placa] = p
    print(f"{placa}: {h} h → ${p}")

for p in dic_cobros.values():
    total_dia += p

print(f"Total del día: ${total_dia}")
```
