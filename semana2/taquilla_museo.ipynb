Para esta actividad necesito crear una manera en la que ayude a un museo con la compra de los boletos que venden, ya que se equivocan mucho y mi programa tiene que ayudar a que no hayan errores en sus calculos ni en los cobros que realizan.

```python
PRECIO_NINO = 0.0
PRECIO_JOVEN = 30.0
PRECIO_ADULTO = 45.0

DESC_ADULTO_MAYOR = 0.12 
DESC_ESTUDIANTE = 0.10    
DESC_PROFESOR = 0.10     
DESC_NINGUNO = 0.0        

EDAD_MIN_VALIDA = 0
EDAD_MAX_VALIDA = 110
EDAD_JOVEN_LIMITE = 3
EDAD_ADULTO_LIMITE = 18
EDAD_MAYOR_LIMITE = 60

print("=== TAQUILLA DEL MUSEO ===")

while True:
    total_visitantes_cobrados = 0
    total_descuentos_acumulados = 0.0
    total_recaudado = 0.0

    while True:
        print()
        entrada = input("Edad del visitante, escribe 'fin' si no agregará mas visitantes: ")
        
        if entrada == 'fin':
            print("Se cierra la taquilla, adiós")
            break
            
        edad = int(entrada)

        if edad < EDAD_MIN_VALIDA or edad > EDAD_MAX_VALIDA:
            print("Una edad negativa o mayor de 110 no es valida")
            continue

        if edad < EDAD_JOVEN_LIMITE:
            precio_base = PRECIO_NINO
            tipo_visitante = "Menor de 3 años"
            monto_descuento = 0.0
            
        elif edad < EDAD_ADULTO_LIMITE:
            precio_base = PRECIO_JOVEN
            tipo_visitante = "Joven de 3-17 años"
            monto_descuento = 0.0
            
        else: 
            precio_base = PRECIO_ADULTO
            
            print("Tipo: 1) Estudiante  2) Profesor  3) Ninguno")
            opcion = input("Seleccione una opcion del 1 al3): ")
            
            if opcion == '1':
                tipo_visitante = "Estudiante"
                monto_descuento = precio_base * DESC_ESTUDIANTE
            elif opcion == '2':
                tipo_visitante = "Profeesor"
                monto_descuento = precio_base * DESC_PROFESOR
            else:
                tipo_visitante = "Ninguno"
                monto_descuento = precio_base * DESC_NINGUNO
            
            if edad >= EDAD_MAYOR_LIMITE:
                tipo_visitante = tipo_visitante + " (Adulto Mayor)"
                descuento_adulto_mayor = precio_base * DESC_ADULTO_MAYOR
                if descuento_adulto_mayor > monto_descuento:
                    monto_descuento = descuento_adulto_mayor

        monto_paga = precio_base - monto_descuento
        
        total_visitantes_cobrados += 1
        total_descuentos_acumulados += monto_descuento
        total_recaudado += monto_paga

        print("=== LINEA DE DETALLE ===")
        print(f" Visitante {total_visitantes_cobrados}", "| Edad:", edad, "|", tipo_visitante, "| Base: $", 
               format(precio_base, ".2f"), "| Descuento: $", format(monto_descuento, ".2f"), 
               "| A pagar: $", format(monto_paga, ".2f"))

    print()
    print("=== CORTE DE CAJA ===")
    print(f"Visitantes cobrados: {total_visitantes_cobrados}")
    print(f"Total en descuentos: ${total_descuentos_acumulados:.2f}")
    print(f"TOTAL A PAGAR: ${total_recaudado:.2f}")


    respuesta_final = input("Quires volver a hacer el proceso? sino escribe 'fin' para terminar: ")
    if respuesta_final == 'fin':
        print("Gracias por comprar boletos, adiós")
        break
```
