```python
matriz = []

for fila in range(3):
    lista = []
    for columna in range(4):
        lista.append(fila + columna)
    matriz.append(lista)

print(matriz)
```

```python
ventas = [[100, 200, 150], [80, 90, 120]]

for i in range(len(ventas)):
    total = sum(ventas[i])
    print("Vendedor", i + 1, ":", total)
```

```python
ventas = [[100, 200, 150], [80, 90, 120]]

for columna in range(3):
    total = 0
    for fila in range(2):
        total = total + ventas[fila][columna]
    print("Día", columna + 1, ":", total)
```

```python
tablero = [
    ["X", "O", "X"],
    [" ", "X", "O"],
    ["O", " ", "X"]]

for fila in tablero:
    print(" | ".join(fila))

    if fila != tablero[2]:
        print("---------")

if tablero[0][0] == "X" and tablero[1][1] == "X" and tablero[2][2] == "X":
    print("¡X gana por la diagonal!")
```

```python
def precio_final(precio, descuento):
    return precio - precio * descuento

boleto = precio_final(1450, 0.10)
print(f"pagas ${boleto}")
```
