```python
ventas = [[100, 200, 150], [80, 90, 120]]

for columna in range(3):
    total = 0
    for fila in range(2):
        total = total + ventas[fila][columna]
    print("Día", columna + 1, ":", total)
```

```python
tablero = [
    ["X", "O", "X"],
    [" ", "X", "O"],
    ["O", " ", "X"]]

for fila in tablero:
    print(" | ".join(fila))

    if fila != tablero[2]:
        print("---------")

if tablero[0][0] == "X" and tablero[1][1] == "X" and tablero[2][2] == "X":
    print("¡X gana por la diagonal!")
```

```python
def duracion_total(canciones):
    total = 0
    for duracion in canciones:
        total = total + duracion
    return total

canciones = [1.70, 3.45, 4.20, 9.30]

print("La duración total es:", duracion_total(canciones))
```

```python
def promedio_calificaciones(calificaciones):
    total = 0
    cantidad = 0

    for calificacion in calificaciones:
        total = total + calificacion
        cantidad = cantidad + 1

    promedio = total / cantidad
    return promedio

calificaciones = [8, 9, 10, 7, 8]

print("El promedio es:", promedio_calificaciones(calificaciones))
```

```python
def promedio_calificaciones(calificaciones):
    total = 0
    cantidad = 0

    for calificacion in calificaciones:
        total = total + calificacion
        cantidad = cantidad + 1

    return total / cantidad


alumnos = {
    "Alumno 1": [8, 9, 10],
    "Alumno 2": [7, 8, 9],
    "Alumno 3": [10, 9, 10],
    "Alumno 4": [6, 8, 9]
}

for nombre in alumnos:
    promedio = promedio_calificaciones(alumnos[nombre])
    print(nombre, "tiene un promedio de:", promedio)
```
