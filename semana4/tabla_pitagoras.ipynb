- Mi programa  construye una tabla de Pitágoras de 10x10 con listas anidadas con dos ciclos for y append(). Ya despues, imprime la matriz completa en formato ordenado y le pide al usuario que ingrese un renglón y una columna, y mediante un ciclo while se asegura de que los datos estén dentro del rango permitido para que de como salida la multilicacion que se espera

Parte A — Construir la matriz

- Construye la tabla de Pitágoras de 10 × 10 como lista de listas, usando dos for anidados y append().
- Aquí sí puedes usar *: cada celda es fila * columna.

```python
tabla = []
for i in range(1, 11):
    fila = []
    for j in range(1, 11):
        fila.append(i * j) 
    tabla.append(fila)
```

Parte B — Mostrarla: función que NO regresa valor

- Define imprimir_tabla(tabla), una función sin return, que imprima la matriz completa.
- La salida debe verse sin corchetes ni comas, con los números separados por tabulaciones (end="\t").

```python
def imprimir_tabla(tabla):
    for fila in tabla:
        for elemento in fila:
            print(elemento, end="\t") 
        print()
```

Parte C — Consultarla: función que SÍ regresa valor

- Define consultar_producto(tabla, renglon, columna), una función que regresa con return el producto consultando la matriz.
- Restricción: dentro de consultar_producto está prohibido usar *. El resultado sale de tabla[renglon - 1][columna - 1].

```python
def consultar_producto(tabla, renglon, columna):
    return tabla[renglon - 1][columna - 1]

imprimir_tabla(tabla)
```

Parte D — Programa principal

- Pide al usuario dos factores (renglón y columna) con input() e int().
- Valida que cada factor esté entre 1 y 10 con un while: si está fuera de rango, muestra un aviso y vuelve a pedirlo.
- Llama a consultar_producto, guarda lo que regresa en una variable y muestra el resultado, por ejemplo: 7 x 8 = 56.

```python
renglon = int(input("Ingresa el renglón que quieres multiplicar: "))  
columna = int(input("Ingresa la columna por la que desees multiplicar: "))

while renglon < 1 or renglon > 10 or columna < 1 or columna > 10:
    print("ERROR, escribe  un número entre 1 y 10")  
    renglon = int(input("Ingresa el renglón que quieres multiplicar: "))
    columna = int(input("Ingresa la columna por la que desees multiplicar: "))

resultado = consultar_producto(tabla, renglon, columna)  
print(f"{renglon} x {columna} = {resultado}")
```

Código completo

```python
tabla = []
for i in range(1, 11):
    fila = []
    for j in range(1, 11):
        fila.append(i * j) 
    tabla.append(fila)

def imprimir_tabla(tabla):
    for fila in tabla:
        for elemento in fila:
            print(elemento, end="\t") 
        print()

def consultar_producto(tabla, renglon, columna):
    return tabla[renglon - 1][columna - 1]  

imprimir_tabla(tabla)

renglon = int(input("Ingresa el renglón que quieres multiplicar: "))  
columna = int(input("Ingresa la columna por la que desees multiplicar: "))

while renglon < 1 or renglon > 10 or columna < 1 or columna > 10:
    print("ERROR, escribe  un número entre 1 y 10")  
    renglon = int(input("Ingresa el renglón que quieres multiplicar: "))
    columna = int(input("Ingresa la columna por la que desees multiplicar: "))

resultado = consultar_producto(tabla, renglon, columna)  
print(f"{renglon} x {columna} = {resultado}")  
```
